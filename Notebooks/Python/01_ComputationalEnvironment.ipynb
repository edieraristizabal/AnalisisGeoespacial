## Iniciar Jupyter Lab

La principal herramienta computacional del curso es [Jupyter Lab](http://jupyter.org/), una interfaz web interactiva para desarrollar código que sigue el concepto conocido como REPL (Read — Evaluate — Print — Loop: leer, evaluar, imprimir, repetir), muy usado por los científicos de datos. La gran ventaja de un entorno REPL es que permite desarrollar el código gradualmente, ejecutando comando por comando y revisando los resultados. Además, se puede mantener el texto explicativo junto al código y ver los resultados en el mismo entorno, sin alternar entre la línea de comandos y otras aplicaciones, como visores de imágenes. Los notebooks son una forma cómoda de entrelazar texto, código y los resultados que produce en un solo archivo que se puede compartir, editar y modificar. Puede pensar en los notebooks como el documento de Word de los científicos de datos.

Un notebook es un solo archivo que guarda texto narrativo, código y los resultados del código. Guardar en un solo archivo el trabajo narrativo y el computacional permite registrar y documentar todo el flujo de trabajo en un mismo lugar, sin recurrir a medios auxiliares (como una libreta de papel). Una segunda característica es que permiten trabajar de forma interactiva: el trabajo computacional moderno se beneficia de poder probar, fallar, ajustar e iterar rápidamente hasta encontrar una solución, y los notebooks encarnan esa cualidad. Ya sea que el cálculo se haga en un portátil o en un centro de datos, los notebooks ofrecen la misma interfaz interactiva, lo que reduce la carga cognitiva de escalar. Tercero, los notebooks son interoperables: el formato está diseñado para registrar y compartir el trabajo computacional, pero no necesariamente para otras etapas del ciclo de investigación, así que para ampliar sus posibilidades se pueden convertir fácilmente a otros formatos. Por ejemplo, aunque se necesita una aplicación específica para abrir y editar la mayoría de los formatos de notebook, no se necesita software adicional para convertirlos en archivos pdf que se puedan leer, imprimir y anotar sin software técnico.

Un notebook de Jupyter es un archivo de texto plano con extensión **.ipynb**, fácil de mover, sincronizar y seguir en el tiempo. Internamente está estructurado como un documento de texto con JavaScript Object Notation (JSON) que registra el estado del notebook, así que también se integra bien con muchas tecnologías web modernas. El elemento básico de un notebook se llama celda: bloques de contenido que contienen texto o código. De hecho, un notebook puede verse como una colección ordenada de celdas, que pueden ser de dos tipos: texto y código.

## Agregar el entorno virtual a Jupyter

Jupyter Notebook garantiza que el kernel de IPython esté disponible, pero hay que agregar manualmente un kernel con otra versión de Python o con un entorno virtual. Primero, asegúrese de que su entorno esté activado con conda activate myenv. Luego instale ipykernel, que provee el kernel de IPython para Jupyter:

```
pip install --user ipykernel
```

Luego puede agregar el entorno virtual a Jupyter escribiendo:

```
python -m ipykernel install --user --name=myenv
```

Si borra el entorno virtual, también querrá quitarlo de Jupyter. Primero veamos qué kernels hay disponibles; se pueden listar con:
```
jupyter kernelspec list
```
Esto debería mostrar los kernels de Jupyter. Para desinstalar el kernel, escriba:
```
jupyter kernelspec uninstall myenv
```

## Contenedor de Docker

Los contenedores son una versión liviana de una máquina virtual, es decir, de un programa que permite ejecutar todo un sistema operativo de forma aislada sobre otro sistema operativo. Los contenedores permiten encapsular un entorno (o plataforma) completo en un formato fácil de transferir y reproducir en distintos contextos computacionales. La tecnología de contenedores más popular hoy es Docker, y apenas se empiezan a explorar las posibilidades que ofrece para construir infraestructura transparente y transferible para la ciencia de datos.

Hay varias formas de describirlo, desde muy técnicas hasta más intuitivas; aquí nos enfocaremos en una comprensión general más que en los detalles técnicos de su magia. Un contenedor puede verse como una "caja" que incluye todo lo necesario para ejecutar cierto conjunto de software. ### Control flow (a.k.a. `for` loops and `if` statements)

Aunque esta no pretende ser una introducción completa a la programación o a Python en general (para eso vea las referencias, en particular el [libro](http://www.greenteapress.com/thinkpython/thinkpython.html) de Allen Downey), es importante conocer dos bloques básicos de casi cualquier programa: los ciclos `for` y las instrucciones `if`. Es posible que nunca los necesite en este curso, porque todo lo que se usa se basa en métodos y funciones existentes, pero siempre es útil saber que existen y reconocerlos. También pueden ser muy útiles cuando se necesita alguna funcionalidad adicional a la de los métodos estándar. Sin más preámbulos, veamos las dos herramientas más relevantes de la programación.

* Ciclos `for`

Permiten repetir una acción o tarea sobre una secuencia. Por ejemplo, puede imprimir su nombre diez veces sin tener que escribirlo cada vez:

````python
for i in np.arange(10):
    print('my name')
````

Note algunas características del ciclo:

1. Se recorre *una* secuencia, en este caso la secuencia de diez números creada con `np.arange(10)`.
1. En cada paso, es decir, para cada elemento de la secuencia, se repite una acción; aquí se imprime el mismo texto, `my name`.
1. Aunque no se usa en este ciclo, cada elemento recorrido se puede usar dentro del ciclo. Puede ser irrelevante, como en el ciclo anterior, o muy útil, según el contexto. Por ejemplo, vea un caso en el que se usa el valor de la secuencia en cada paso:

````python
for i in np.arange(10):
    print("I am at step ", i)
````

Una nota más: por convención llamamos `i` al elemento de la secuencia, pero podría tener cualquier nombre. De hecho, en muchos casos los nombres más significativos hacen el código mucho más legible. Por ejemplo, el ciclo anterior se podría reescribir así:

````python
for step in np.arange(10):
    print("I am at step ", step)
````

* Instrucciones `if`

Acabamos de ver que los ciclos `for` permiten repetir una acción sobre una secuencia. Las instrucciones `if` permiten seleccionar o restringir esas acciones a los casos que cumplen las condiciones indicadas.

Por ejemplo, en los ciclos anteriores podríamos querer imprimir solo los números impares y saltar los pares:

````python
for i in np.arange(10):
    if i%2:
        print(i)
````

Por ahora ignore la parte `i%2`; solo recuerde que es una forma que tiene Python de verificar si un número es impar. Lo importante de este ciclo, frente al anterior más sencillo, es que usamos una instrucción `if` para seleccionar solo los candidatos que cumplen la condición. Es decir, recorremos cada número de la secuencia de cero a nueve (`for i in np.arange(10)`) y verificamos si es par o impar (`if i%2`). Si cumple la condición, es decir, si es impar, lo imprimimos en pantalla.

Una instrucción `if` completa también permite hacer algo cuando la condición original no se cumple. Esto se llama instrucción "if-else". Por ejemplo, se puede pensar en un ciclo que imprima el tipo de cada número de una secuencia:

````python
for i in np.arange(10):
    # Verificar si es impar
    if i%2:
        print(i, ' is odd')
    # Si no es impar (es par), hacer lo siguiente
    else:
        print(i, ' is even')
````

### Functions

La última parte de este recorrido rápido por Python tiene que ver con las funciones, o más propiamente, los métodos. La motivación es que, hasta ahora, solo hemos visto cómo crear código de Python que, si queremos ejecutar de nuevo en otro lugar, hay que copiar y pegar completo. Sin embargo, como veremos con más detalle en el curso, una de las razones principales para usar Python en el análisis de datos, en lugar de una interfaz gráfica de apuntar y hacer clic como SPSS, es que el código se puede reutilizar y los análisis se pueden volver a ejecutar fácilmente. Los métodos ayudan a lograrlo encapsulando fragmentos de código que hacen una tarea particular y dejándolos disponibles para ser llamados.

Ya hemos *usado* métodos: al llamar `np.arange` usamos uno de ellos. Ahora veremos cómo *crear* un método propio que haga la tarea específica que queremos. Por ejemplo, creemos un método que reproduzca el primer ciclo que hicimos antes:

````python
def run_loop():
    for i in np.arange(10):
        print(i)
    return None
````

Ya con este método pasan varias cosas interesantes:

* Primero, note cómo se define que un fragmento de código es un método y no Python común: se usa `def` seguido del nombre de la función (elegimos `run_loop`, pero podría ser cualquiera).
* Segundo, se agrega `()` después del nombre y se termina la línea con dos puntos (`:`). Esto es necesario y permite especificar los requisitos de la función (ver abajo).
* Tercero, todo lo que está dentro de una función debe estar indentado. Es una propiedad central de Python y, aunque a algunas personas les parece extraño, mejora mucho la legibilidad.
* Cuarto, el código que hace la tarea que queremos, imprimir la secuencia de números, está dentro de la función igual que estaba afuera, solo que indentado.
* Quinto, el método termina con una línea que empieza por `return`. En este caso la sigue `None`, pero esto cambiará a medida que los métodos se vuelvan más sofisticados. En esencia, es la parte del método donde se indica qué elementos debe devolver y guardar para usarlos después.

Teniendo en cuenta estos elementos, veamos cómo se puede *llamar* el método y, por lo tanto, ejecutar el código que contiene:

````python
run_loop()
````

Es la misma forma en que llamamos antes a `np.arange`. Note que no se incluyen los dos puntos (`:`), sino simplemente el nombre del método seguido de los paréntesis.

Este método tiene una complejidad limitada: no requiere ninguna entrada, y el código produce una salida (lo que imprime) pero no la guarda en ningún lado. El resto de esta sección flexibiliza estos dos aspectos para construir métodos más complejos, pero también más útiles.

Primero, se pueden especificar "argumentos" que modifiquen el comportamiento del método. Recuerde que llamamos `np.arange` con un número que indicaba la longitud de la secuencia que queríamos. Podemos hacer lo mismo en nuestra propia función. Lo principal es que los argumentos deben ser variables, no valores particulares. Veamos un ejemplo modificado del método:

````python
def run_loopX(x):
    for i in np.arange(x):
        print(i)
    return None
````

Reemplazamos la longitud fija de la secuencia (10) por una variable llamada `x`, que permite especificar *cualquier valor que queramos* al llamar el método:

````python
run_loopX(3)
````

````python
run_loopX(2)
````

Otra forma de dar más flexibilidad a un método es permitir que devuelva un resultado del cálculo. En los ejemplos anteriores, la función hace un cálculo (imprimir valores en pantalla) pero no devuelve ningún valor. Esto contrasta, por ejemplo, con `np.arange`, que sí devuelve un resultado, la secuencia de valores:

````python
a = np.arange(10)
````

````python
a
````

Nuestra función no guarda nada:

````python
b = run_loopX(3)
````

````python
b
````

Podemos cambiar esto con la última línea del método. Por ejemplo, supongamos que queremos devolver una secuencia tan larga como la serie de números que imprimimos en pantalla. El método sería:

````python
def run_loopXout(x):
    for i in np.arange(x):
        print(i)
    return np.arange(x)
````

Note la diferencia principal: en lugar de devolver `None`, le decimos a Python que devuelva una secuencia de la misma longitud que la usada en el ciclo. Hay una forma alternativa más eficiente de escribir este método: asignar la secuencia a un nuevo objeto y usarlo cuando sea necesario. Los resultados son exactamente los mismos, pero se hacen menos cálculos y, sobre todo, se reducen las posibilidades de cometer errores.

````python
def run_loopXout(x):
    seq = np.arange(x)
    for i in seq:
        print(i)
    return seq
````

Cualquiera de estas dos nuevas versiones del método devuelve un resultado:

````python
a = run_loopX(3)
b = run_loopXout(3)
````

````python
a
````

````python
b
````

La ventaja de los métodos, frente al código directo, es que obligan a pensar de forma modular y ayudan a identificar exactamente qué hay que hacer, en qué orden y qué se necesita. Encapsular estas piezas básicas de funcionalidad en métodos permite escribir las cosas una vez y usarlas con flexibilidad en todas partes, lo que ahorra tiempo (y dolores de cabeza) a largo plazo.

Una nota final sobre las funciones. Es importante que, al crear una función, incluya documentación sobre lo que espera recibir, lo que hace y lo que devuelve. Aunque hay muchas formas de hacerlo, la convención típica es la siguiente:

````python
def run_loopXout(x):
    """
    Print out the values of a sequence of certain length
    ...
    
    Arguments
    ---------
    x     : int
            Length of the sequence to be printed out
    
    Returns
    -------
    seq   : np.array
            Sequence of values printed out
    """
    seq = np.arange(x)
    for i in seq:
        print(i)
    return seq
````

La documentación, como cualquier cadena de texto, aparece resaltada en rojo en un notebook. Veamos la estructura y los componentes de una documentación bien hecha (también llamada "docstring"):

* Está encerrada entre triples comillas (`"""`).
* Empieza con una descripción breve de lo que hace el método. Cuanto más corta y concisa, mejor.
* Tiene una sección llamada "Arguments" que enumera los elementos que espera la función. 
* Cada argumento se enumera seguido de su tipo. En este caso es un objeto `x` que, según se indica, debe ser un entero.
* Después de los argumentos hay otra sección que indica qué devuelve la función y de qué tipo es el resultado.

Documentar así es muy útil para recordar qué hace una función, pero también obliga a escribir código más claro. Además, si se documenta de esta forma, la documentación se puede consultar con los sistemas estándar `help` o `?` vistos antes:

````python
run_loopXout?
````

````python
help(run_loopXout)
````

Google Earth Engine (GEE) es una plataforma en la nube que permite el análisis científico y la visualización de conjuntos de datos geoespaciales a gran escala. Google la lanzó en 2010 como un sistema propietario. Actualmente está disponible como un servicio gratuito para cargas de trabajo pequeñas y medianas, con un modelo de negocio similar al de los demás servicios en la nube de la compañía. La plataforma se construye sobre un conjunto de tecnologías de la infraestructura de Google, como el sistema de gestión de clústeres a gran escala (Borg), las bases de datos distribuidas (Bigtable y Spanner), el sistema de archivos distribuido (Colossus) y el marco de ejecución de flujos en paralelo FlumeJava

Google Earth Engine permite ejecutar algoritmos sobre imágenes y vectores georreferenciados almacenados en la infraestructura de Google. La API de Google Earth Engine ofrece una librería de funciones que se pueden aplicar a los datos para visualizarlos y analizarlos. El catálogo público de Earth Engine contiene una gran cantidad de imágenes y conjuntos de datos vectoriales de acceso público, y los usuarios también pueden crear recursos privados en sus carpetas personales.

GEE ofrece un catálogo con un gran repositorio de datos geoespaciales, que incluye imágenes ópticas de muchos satélites y sistemas aéreos, variables ambientales, pronósticos del tiempo y del clima, cobertura del suelo y conjuntos de datos socioeconómicos y topográficos. Antes de ponerse a disposición, estos datos se preprocesan, lo que permite un acceso eficiente y elimina muchas barreras de la gestión de datos.

GEE usa cuatro tipos de objetos para representar los datos que se pueden manipular con su API. El tipo Image representa datos ráster de una o más bandas, cada una con nombre, tipo de dato, escala y proyección. Una pila o serie de tiempo de imágenes se representa con el tipo ImageCollection. Los datos vectoriales se representan con el tipo Feature, formado por una geometría (punto, línea o polígono) y una lista de atributos. El tipo FeatureCollection representa grupos de Features relacionados y ofrece funciones para manipularlos, como ordenar, filtrar y visualizar. GEE solo ofrece interfaces de programación para el procesamiento por píxel. El resultado de un procesamiento en GEE se puede ver en el IDE web o guardar en uno de tres servicios de la compañía: Drive, Cloud Storage o Assets. GEE usa un servidor de teselas para poner los datos a disposición de la interfaz web de forma eficiente.

GEE ofrece una API de JavaScript y una de Python para gestionar y analizar datos. Para la versión de JavaScript también hay un entorno de desarrollo integrado web [(IDE)](https://code.earthengine.google.com), donde el usuario accede fácilmente a los datos, las aplicaciones y la visualización en tiempo real de los resultados. La API de Python está disponible como módulo y tiene una estructura similar a la de JavaScript.

#### El Code Editor 
El Code Editor es un entorno interactivo para desarrollar aplicaciones de Earth Engine. El panel central es un editor de código JavaScript. Encima del editor hay botones para guardar el script, ejecutarlo y limpiar el mapa; el botón Get Link genera una URL única del script en la barra de direcciones. El mapa del panel inferior contiene las capas agregadas por el script, y arriba hay un buscador de conjuntos de datos y lugares. El panel izquierdo contiene ejemplos de código, los scripts guardados, una referencia de la API con buscador y un administrador de recursos para datos privados. El panel derecho tiene un inspector para consultar el mapa, una consola de salida y un administrador de tareas de larga duración. El botón de ayuda, arriba a la derecha, tiene enlaces a esta guía y a otros recursos de ayuda. Más información en la [guía del Code Editor](https://developers.google.com/earth-engine/guides/playground).

El Code Editor de Earth Engine (EE), en [code.earthengine.google.com](https://code.earthengine.google.com/), es un IDE web para la API de JavaScript de Earth Engine. Sus funciones están diseñadas para que desarrollar flujos de trabajo geoespaciales complejos sea rápido y sencillo. Tiene los siguientes elementos (ilustrados en la figura 1):

- Editor de código JavaScript
- Mapa para visualizar conjuntos de datos geoespaciales
- Documentación de referencia de la API (pestaña Docs)
- Administrador de scripts basado en Git (pestaña Scripts)
- Salida de consola (pestaña Console)
- Administrador de tareas (pestaña Tasks) para consultas de larga duración
- Consulta interactiva del mapa (pestaña Inspector)
- Búsqueda en el archivo de datos o en los scripts guardados
- Herramientas de dibujo de geometrías

Los pasos siguientes muestran cómo abrir Earth Engine y ejecutar un script que muestra una imagen. Para mejores resultados, conviene instalar la última versión de Chrome, el navegador de Google.

1. Abra el Code Editor de Earth Engine en [code.earthengine.google.com](https://code.earthengine.google.com/). Si aún no lo ha hecho, debe habilitar el acceso iniciando sesión con una cuenta de Google registrada.
2. Vaya a la pestaña Scripts, en el extremo izquierdo del Code Editor. Allí encontrará una colección de scripts de ejemplo que acceden a datos de Earth Engine, los muestran y los analizan.
3. En “Image Collection”, seleccione el ejemplo “Filtered Composite”. El script aparecerá en la consola central. Presione el botón Run para ejecutar el 4. script. El ejemplo Filtered Composite selecciona imágenes de Landsat 7 que intersecan o están dentro de los límites de Colorado y Utah, y muestra una composición en color verdadero de las imágenes seleccionadas. Los ejemplos presentan métodos de uso común, como filter(), clip() y Map.addLayer().

El Code Editor tiene varias funciones para aprovechar la API de Earth Engine. En la pestaña Scripts puede ver scripts de ejemplo o guardar los suyos. Con la pestaña Inspector puede consultar los objetos del mapa. Puede mostrar y graficar resultados numéricos con la API de visualización de Google, y compartir con colegas una URL única de su script con el botón Get Link. Los scripts que desarrolla en el Code Editor se envían a Google para procesarlos, y las teselas de mapa y/o mensajes generados se devuelven para mostrarlos en el mapa y/o en la pestaña Console. Para usar el Code Editor solo se necesita un navegador web (use Google Chrome para mejores resultados) y conexión a internet. #### La API de Python de Earth Engine
La API de Python de Earth Engine se distribuye como paquete de conda-forge en: https://anaconda.org/conda-forge/earthengine-api, y se instala con el comando conda install. Antes de instalarla, sin embargo, cree un entorno de conda específico para Earth Engine. Instalar la API en su propio entorno garantiza que ni ella ni sus dependencias causen problemas de versiones con el entorno base o con cualquier otro entorno que haya configurado, y viceversa. Para más información sobre la gestión de entornos de conda, visite este [sitio](https://docs.conda.io/projects/conda/en/latest/user-guide/concepts/environments.html).

1. Active el entorno base de conda, si no lo está.
2. Cree un entorno virtual de conda para la API de Earth Engine.

```
conda create --name ee
```
3. Active el entorno ee de conda.

```
conda activate ee
```
4. Instale la API en el entorno ee. Verifique que (ee) aparezca al inicio de la línea de comandos, lo que indica que está trabajando en el entorno ee.

```
conda install -c conda-forge earthengine-api
```
Antes de usar la API de Earth Engine o la herramienta de línea de comandos earthengine, hay que hacer una autenticación única que autoriza el acceso a Earth Engine en nombre de su cuenta de Google. Para autenticarse, use el comando authenticate de la herramienta earthengine.

Dentro del entorno ee, ejecute el siguiente comando y siga las instrucciones que aparecen. Se le dará una URL que genera un código de autorización al aceptar. Copie el código y escríbalo en la línea de comandos.
```
earthengine authenticate
```
Al ingresar el código, se guarda un token de autorización en un archivo de credenciales. Los usos posteriores del comando ee.Initialize() de la API y de la herramienta earthengine buscarán este archivo para autenticarse. Si quiere revocar la autorización, basta con borrar el archivo de credenciales.
```
ee.Initialize()
```
Cada vez que quiera usar ee:
```
import ee

# Inicializar el módulo de Earth Engine.
ee.Initialize()

# Imprimir los metadatos de un modelo digital de elevación.
print(ee.Image('USGS/SRTMGL1_003').getInfo())
```

#### Sentinel Hub
[Sentinel Hub](https://www.sentinel-hub.com/) (SH) es una plataforma desarrollada por Sinergise que ofrece servicios de acceso y visualización de datos Sentinel. Es una plataforma privada de acceso público (https://www.sentinel-hub.com). A diferencia de Google Earth Engine, SH limita el acceso a sus funciones según distintos planes de pago. El plan gratuito solo permite ver, seleccionar y descargar datos crudos. El acceso pago permite acceder a los datos mediante protocolos OGC y una API específica, procesar datos, acceder desde aplicaciones móviles, tener límites de recursos más altos y soporte técnico.

<center><img src="https://www.sinergise.com/sites/default/files/styles/content_image/public/field/image/sent_thumbnail4.jpg?itok=ESzvtSJZ"></center>

[SH](https://www.sentinel-hub.com/explore/) ofrece dos aplicaciones web de sensores remotos muy conocidas ([EO Browser](https://apps.sentinel-hub.com/eo-browser/?zoom=10&lat=41.9&lng=12.5&themeId=DEFAULT-THEME) y [Sentinel Playground](https://apps.sentinel-hub.com/sentinel-playground/)). Sentinel Playground es un visor web sencillo para ver rápidamente en línea el archivo de imágenes de Sentinel-1, Sentinel-2, Landsat 8 y MODIS; EO Browser aprovecha potentes funciones de scripts, permite explorar 12 temas, crear secuencias temporales (timelapses) y descargar imágenes de alta resolución, todo con el archivo completo de datos de SH.

<center><img src="http://www.gisandbeers.com/wp-content/uploads/2017/08/Sentinel-Playground.jpg"></center>

EO Browser permite explorar y comparar imágenes a resolución completa de todas las fuentes de datos disponibles. Basta con ir al área de interés, seleccionar el rango de tiempo y la cobertura de nubes deseados, y revisar los datos resultantes en el navegador. Se pueden probar distintas visualizaciones o crear las propias, descargar imágenes de alta resolución y crear secuencias temporales. Las imágenes de satélite se pueden visualizar según la configuración que el usuario quiera; ya hay varias visualizaciones preparadas con leyendas y descripciones, como color verdadero, falso color, NDVI, EVI, etc. Con la opción Custom se puede elegir cualquier combinación de bandas y hacer una composición arrastrando las bandas a los canales RGB. La herramienta de índices permite crear rápidamente índices de sensores remotos y controlar su visualización, arrastrando bandas a la ecuación. La función de scripts personalizados es una herramienta poderosa para visualizar datos satelitales: con JavaScript se tiene control total sobre la visualización, lo que permite hacer cálculos, usar operadores lógicos y condiciones, fusión de datos, scripts multitemporales, etc. Además, en el panel de efectos se puede modificar la imagen ajustando la intensidad de los tres canales de color, el contraste (gain) y la luminancia (gamma).

<center><img src="https://www.sentinel-hub.com/img/glacier.jpg" width="800"></center>

Para obtener una imagen se utiliza el siguiente script:

```
//color verdadero mejorando brillo (*) y contraste (-), y trasnparencia como cuarto valor (0.5)
return[(2.5*B04)-0.1,(2.5*B03)-0.1,(2.5*B02)-0.1,0.5]
```
Para calcular el NDVI y generar categorias, se puede utilizar:
```
//NDVI categorizado
var NDVI = (B08-B04)/(B08+B04)

if (NDVI<0.2){
	return[1,1,1]
}

if (NDVI<0.4){
	return[0,1,0]
}

if (NDVI<0.7){
	return[0,0.5,0]
}

else{
	return[0,0,0]
}
```
