```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Definir la población: Distribución Exponencial (ej. Permeabilidad)
np.random.seed(42)
poblacion = np.random.exponential(scale=5.0, size=10000)
media_poblacion = np.mean(poblacion)

# 2. Simulación del TLC: Tomar 1000 muestras de tamaño n
n_muestras = 1000
tamano_muestra = 100
medias_muestrales = [np.random.choice(poblacion, size=tamano_muestra).mean() for _ in range(n_muestras)]
media_de_medias = np.mean(medias_muestrales)

# 3. Graficar
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Población Original
sns.histplot(poblacion, kde=True, ax=axes[0], color='skyblue', stat="density")
axes[0].axvline(media_poblacion, color='red', linestyle='--', linewidth=2, label=f'Media: {media_poblacion:.3f}')
axes[0].set_title('Población Original\n(Datos Crudos del Terreno)')
axes[0].legend()

# Medias Muestrales
sns.histplot(medias_muestrales, kde=True, ax=axes[1], color='salmon', stat="density")
axes[1].axvline(media_de_medias, color='red', linestyle='--', linewidth=2, label=f'Media de Medias: {media_de_medias:.3f}')
axes[1].set_title(f'Distribución de Medias (n={tamano_muestra})\n(Lo que el geólogo estima)')
axes[1].legend()

```
