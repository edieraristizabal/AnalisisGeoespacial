# Web spatial tools

## Landsat en AWS
Los datos de Landsat 8 están disponibles para cualquier persona a través de [Amazon S3](https://docs.opendata.aws/landsat-pds/readme.html), desde el inicio de la captura de imágenes. Las nuevas escenas de Landsat 8 se publican cada día, a menudo pocas horas después de su producción.

## [EOX COG-Explorer](https://eox.at/2018/05/explore-landsat-8-scenes-with-cog-explorer/)
Esta aplicación permite inspeccionar visualmente GeoTIFF optimizados para la nube (COG) individuales, como las escenas del archivo de Landsat-8 en [Amazon S3](https://geotiffjs.github.io/cog-explorer/#long=16.370&lat=48.210&zoom=5&scene=&bands=&pipeline=)

## [Land viewer](https://eos.com/products/landviewer/)
Land Viewer es una herramienta de imágenes de observación satelital que permite buscar y procesar datos satelitales al instante y obtener información valiosa para resolver problemas reales

## Spectator
[Spectator Earth](https://app.spectator.earth/) es una aplicación web que ofrece imágenes de observación de la Tierra, principalmente de satélites con datos abiertos como las familias Sentinel y Landsat


## t-SNE
Se creó un mapa mundial alternativo que refleja lo que pasaría con la situación de cada país si cambian ciertos parámetros. Como se trata de visualización de datos y no de geopolítica, no se hace ninguna evaluación de estos datos. Se usó un algoritmo de aprendizaje automático [t-SNE](https://projects.interacta.io/country-tsne/) para procesar los datos, y el usuario puede variar la tasa de desempleo, los costos de salud y otras características.

## Carto
[Carto](https://carto.com/) es una plataforma de computación en la nube de software como servicio (SaaS) que ofrece herramientas de SIG, cartografía web y ciencia de datos espaciales.

## Kepler
[Kepler](https://kepler.gl/) es una herramienta web de alto rendimiento creada por el equipo de visualización de Uber para la exploración visual de conjuntos de datos geoespaciales a gran escala. Está construida sobre deck.gl, otro marco de Uber para el análisis exploratorio visual de grandes conjuntos de datos.

kepler.gl también se puede usar dentro de un [notebook de Jupyter](https://www.analyticsvidhya.com/blog/2020/06/learn-visualize-geospatial-data-jupyter-kepler/#:~:text=Kepler.gl%20is%20a%20high,of%20large%20datasets%20by%20Uber.), lo que la hace muy práctica para toda la comunidad de ciencia de datos.

## [Sentinelhub Playground](https://apps.sentinel-hub.com/sentinel-playground/?source=L8)
Sentinel Hub Playground ofrece un SIG web para explorar imágenes de las plataformas Sentinel-2, Landsat-8 y MODIS. Los datos de Sentinel y Landsat se pueden mostrar en varias combinaciones de bandas e índices espectrales predefinidos, que se generan al instante, y el usuario puede crear combinaciones de bandas personalizadas.

## [Geocoder](https://geocode.localfocus.nl/)
La geocodificación es el proceso de transformar la descripción de una ubicación (un par de coordenadas, una dirección o el nombre de un lugar) en una ubicación sobre la superficie terrestre. Se puede geocodificar una descripción a la vez o muchas a la vez desde una tabla. Las ubicaciones resultantes se obtienen como elementos geográficos con atributos, que se pueden usar para cartografía o análisis espacial.

## RAWGraphs
[RAWGraphs](https://app.rawgraphs.io/) es un marco de visualización de datos de código abierto, creado para que cualquier persona pueda representar visualmente datos complejos con facilidad.

## Palladio
[Palladio](http://hdlab.stanford.edu/palladio/) es una herramienta web muy usada para visualizar datos mediante redes y mapas


## Eastraea - EarthAI
[EarthAI](https://astraea.earth/platform/) es una plataforma integrada de análisis de imágenes diseñada para descubrir, acceder y procesar datos de observación de la Tierra
